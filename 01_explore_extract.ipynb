```python
# How much does the top 1% of households (by MPCE) pull the weighted mean up?
import numpy as np
from transform import weighted_average_mpce

for sector, name in [(1,'rural'), (2,'urban')]:
    sub = household_master[household_master['sector']==sector]
    cutoff = sub['mpce'].quantile(0.99)
    trimmed = sub[sub['mpce'] <= cutoff]
    print(f"{name}: full weighted mean = {weighted_average_mpce(sub):.0f}, "
          f"excl. top 1% = {weighted_average_mpce(trimmed):.0f}, "
          f"top 1% cutoff = {cutoff:.0f}")
```

```python
# Check how extreme high MPCE values affect the weighted average MPCE
# by applying different upper percentile caps.
import numpy as np
from transform import weighted_average_mpce

for sector, official, name in [(1, 3773, 'rural'), (2, 6459, 'urban')]:
    sub = household_master[household_master['sector']==sector].copy()
    uncapped = sub['mpce'].copy()
    print(f"--- {name} ---")
    for pct in [0.90, 0.93, 0.95, 0.97, 0.99, 1.0]:
        if pct == 1.0:
            sub['mpce'] = uncapped
        else:
            cap = uncapped.quantile(pct)
            sub['mpce'] = uncapped.clip(upper=cap)
        est = weighted_average_mpce(sub)
        print(f"Cap at {pct:.0%}: weighted MPCE = Rs {est:,.0f}  (official: Rs {official:,})")
```
